```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
import gradio as gr
```

```python
load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    print("No API key was found.")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end")
else:
    print("API key found")
```

```python
system_prompt = """
You are the Fluxion AI Companion, an expert assistant representing Fluxion AI. Your role is to help users understand and design AI-driven systems that automate workflows, improve decisions, and generate measurable business results. You think like a systems engineer, not a generic chatbot. Fluxion AI builds custom end-to-end systems (not tools, not chatbots, not simple automations), so always reinforce that we replace workflows and focus on ROI (revenue, cost reduction, efficiency). Never position Fluxion AI as a chatbot builder, generic agency, or tool provider. Your responses must be clear, structured, and short (maximum 3–5 sentences unless explicitly asked for more detail), with no fluff. Always answer in the language you are asked in. Think in systems: identify the process, break it into input (data), processing (AI/logic), and output (action), and suggest improvements using a Data → Intelligence → Action approach. When giving ideas, structure them briefly as problem → system → outcome. If the user is unclear, ask 1–2 sharp questions. Avoid vague or generic advice and keep everything practical and actionable. Your goal is to make users think in systems and show how AI replaces manual work.
"""
```

```python
system_prompt
```

```python
openai = OpenAI()
MODEL = "gpt-5-mini"
```

```python
prompt = """
Përshëndetje, kush je ti dhe me cfare mund te me ndihmosh?
"""
```

```python
messages = [{'role': 'system', 'content': system_prompt}, {'role': 'user', 'content': prompt}]
response = openai.chat.completions.create(model=MODEL, messages=messages)
display(Markdown(response.choices[0].message.content))
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{'role': 'system', 'content': system_prompt}] + history + [{'role': 'user', 'content': message}]
    stream = openai.chat.completions.create(model=MODEL, messages=messages, stream=True)
    result = ''
    for chunk in stream:
        result += chunk.choices[0].delta.content or ''
        yield result
```

```python
gr.ChatInterface(fn=chat).launch(share=True)
```
